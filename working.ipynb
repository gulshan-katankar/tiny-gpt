```python
# read it in to inspect it
with open('input.txt', 'r', encoding='utf-8') as f:
  text = f.read()
```

```python
print("length of the dataset at character level : ", len(text))
```

```python
#looking at the first 1000 char
print(text[:1000])
```

```python
#the unique chars
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

so for tokenizing we are keeping it simple here, using a char level token which just maps the char to an integer see the example code below:

```python
#creating a mapping from char to integers

stoi = { ch:i for i,ch in enumerate(chars) }
itos = { i:ch for i,ch in enumerate(chars) }
encode = lambda s: [stoi[c] for c in s] # encoder: take a string, output a list of integers
decode = lambda l: ''. join( [itos[i] for i in l]) # decoder: take a list of integers, output a string

print(encode("hii there"))
print(decode(encode("hii there")))
```

now we will apply this example code to our full dataset

```python
#store the encoded dataset in torch.Tensor

import torch
data = torch.tensor(encode(text), dtype=torch.long)
print(data.shape, data.dtype)
print(data[:1000])
```

the above tensor you see is just a part of the full dataset, we have encoded the chars, changing them into a massive sequence of integers

```python
#split the data
n = int(0.9*len(data)) #first 90% will be train, rest val
train_data = data[:n]
val_data = data[n:]
```
